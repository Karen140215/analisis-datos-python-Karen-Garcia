## 9. Tuplas

Una tupla es una colección ordenada e inmutable.  
Se usa cuando los datos no deberían cambiar, por ejemplo, coordenadas, rangos de operación o configuraciones fijas.

````python
coordenada_robot = (2.5, 4.0)

print("Coordenada:", coordenada_robot)
print("x:", coordenada_robot[0])
print("y:", coordenada_robot[1])
````

### Desempaquetado de tuplas

````python
x, y = coordenada_robot

print("Posición x:", x)
print("Posición y:", y)
````

## 10. Diccionarios

Un diccionario almacena información en pares `clave: valor`.  
Es muy útil para representar registros estructurados.

````python
motor = {
    "id": "M01",
    "tipo": "Motor DC",
    "voltaje_V": 12,
    "corriente_A": 1.8,
    "estado": "Activo"
}

print(motor)
print("Tipo:", motor["tipo"])
print("Estado:", motor["estado"])
````

### Acceso, actualización y creación de claves

````python
motor["temperatura_C"] = 42.5
motor["estado"] = "Precaución"

print(motor)
print("Potencia:", motor["voltaje_V"] * motor["corriente_A"], "W")
````

### Uso de `.get()`

El método `.get()` permite consultar una clave sin producir error si la clave no existe.

````python
print(motor.get("ubicacion", "Ubicación no registrada"))
````

### Lista de diccionarios

En análisis de datos, muchas veces se tienen varios registros. Una forma básica de representarlos es una lista de diccionarios.

````python
registros = [
    {"muestra": 1, "temperatura_C": 35.2, "corriente_A": 1.5},
    {"muestra": 2, "temperatura_C": 38.0, "corriente_A": 1.7},
    {"muestra": 3, "temperatura_C": 41.5, "corriente_A": 2.0},
]

for registro in registros:
    print(registro["muestra"], registro["temperatura_C"], registro["corriente_A"])
````

## 11. Conjuntos

Un conjunto o `set` almacena valores únicos.  
Es útil para eliminar duplicados o identificar categorías diferentes.

````python
sensores = ["temperatura", "corriente", "vibracion", "temperatura", "corriente"]
sensores_unicos = set(sensores)

print("Sensores originales:", sensores)
print("Sensores únicos:", sensores_unicos)
````

### Operaciones con conjuntos

````python
sensores_equipo_A = {"temperatura", "corriente", "vibracion"}
sensores_equipo_B = {"temperatura", "presion", "flujo"}

print("Unión:", sensores_equipo_A | sensores_equipo_B)
print("Intersección:", sensores_equipo_A & sensores_equipo_B)
print("Solo en A:", sensores_equipo_A - sensores_equipo_B)
print("Diferencia simétrica:", sensores_equipo_A ^ sensores_equipo_B)
````
