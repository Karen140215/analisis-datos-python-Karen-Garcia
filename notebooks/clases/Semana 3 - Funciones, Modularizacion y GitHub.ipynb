## 1. ¿Por qué usar funciones en análisis de datos?

En análisis de datos, muchas tareas se repiten:

- Calcular indicadores.
- Validar rangos.
- Limpiar valores.
- Clasificar estados.
- Generar resúmenes.
- Transformar columnas.
- Crear gráficas.

Las funciones permiten organizar estas tareas y reutilizarlas.

````python
# Ejemplo sin función: código repetido

temperatura_1 = 35
temperatura_2 = 44
temperatura_3 = 52

if temperatura_1 < 40:
    estado_1 = "Normal"
elif temperatura_1 < 50:
    estado_1 = "Precaución"
else:
    estado_1 = "Alerta"

if temperatura_2 < 40:
    estado_2 = "Normal"
elif temperatura_2 < 50:
    estado_2 = "Precaución"
else:
    estado_2 = "Alerta"

if temperatura_3 < 40:
    estado_3 = "Normal"
elif temperatura_3 < 50:
    estado_3 = "Precaución"
else:
    estado_3 = "Alerta"

print(estado_1, estado_2, estado_3)
````

El código anterior funciona, pero es repetitivo.  
La solución es crear una función.

````python
def clf_temperatura(temperatura):
  if temperatura < 30:
    estado = "Normal"
  elif temperatura< 50:
    estado = "Precaución"
  else:
    estado = "Alerta"


print(clf_temperatura(35))
print(clf_temperatura(44))
print(clf_temperatura(52))
````

## 3. Parámetros con valores por defecto

Los parámetros con valores por defecto permiten que una función use un valor estándar cuando no se entrega un argumento.

````python
def calcular_energia(potencia_w, tiempo_h=1):
    return potencia_w * tiempo_h

print("Energía en 1 hora:",calcular_energia(24) )
print("Energía en 3 horas:", calcular_energia(24,3))
````

## 4. Funciones con validación básica

Una buena función puede verificar que los datos de entrada sean válidos antes de hacer el cálculo.

````python
def calcular_potencia_segura(voltaje,corriente):
  if voltaje <0 or corriente <0 :
    return None
  return voltaje * corriente

print("Potencia:", calcular_potencia_segura(12, -5))
print("Potencia:", calcular_potencia_segura(12, 5))
````

## 5. Modularización básica

Modularizar significa dividir un problema en partes pequeñas.  
Por ejemplo, para analizar mediciones de un motor, se pueden crear funciones para:

1. Calcular potencia.
2. Clasificar temperatura.
3. Clasificar corriente.
4. Generar un resumen.

````python
def calcular_potencia(voltaje, corriente):
    return voltaje * corriente

def clasificar_temperatura(temp):
    if temp < 40:
        return "Normal"
    elif temp < 50:
        return "Precaución"
    else:
        return "Alerta"

def clasificar_corriente(corriente):
    if corriente < 2.0:
        return "Baja"
    elif corriente <= 3.0:
        return "Normal"
    else:
        return "Alta"

def generar_resumen(voltaje, corriente, temperatura):
    potencia = calcular_potencia(voltaje, corriente)
    estado_temp = clasificar_temperatura(temperatura)
    estado_corriente = clasificar_corriente(corriente)

    resumen = {
        "voltaje_V": voltaje,
        "corriente_A": corriente,
        "temperatura_C": temperatura,
        "potencia_W": potencia,
        "estado_temperatura": estado_temp,
        "estado_corriente": estado_corriente
    }

    return resumen

generar_resumen(12, 2.5, 43)
````

## 6. Comprensión de listas

La comprensión de listas permite crear listas de forma compacta.

Sintaxis general:

```python
nueva_lista = [expresion for elemento in lista]
```

Captura de Pantalla 2022-05-30 a la(s) 10.28.05 a.m..png

````python
temperaturas = [35, 38, 41, 45, 52]

# Forma tradicional
temperaturas_f = []

for temp in temperaturas:
    temp_f = (temp * 9/5) + 32
    temperaturas_f.append(temp_f)

print(temperaturas_f)
````

````python
# Con comprensión de listas
temperaturas_f = [(temp * 9/5) + 32 for temp in temperaturas]
print(temperaturas_f)
````

También se pueden incluir condiciones.

````python
temperaturas_altas = [temp for temp in temperaturas if temp > 45]
print(temperaturas_altas)
````

````python
estados = [clasificar_temperatura(temp) for temp in temperaturas]
print(estados)
````

## 8. Manejo de errores

En análisis de datos pueden aparecer errores por:

- Valores vacíos.
- Tipos de datos incorrectos.
- Divisiones por cero.
- Archivos inexistentes.
- Columnas mal nombradas.
- Datos fuera de rango.

Python permite manejar errores mediante `try` y `except`.

````python
valor = "12.5"
try:
  numero = float (valor)
  print("Numero convertido:", numero)
  print(numero/ valor)
except:
  print("No se puede convertir en numero")
````

## 9. Función robusta para conversión de datos

Una función robusta puede manejar datos incorrectos sin detener todo el análisis.

````python
def convertir_a_float(valor):
  try :
    return float (valor)
  except:
    return None

datos_crudos = ["34.5", "36.0", "", "error", "42.5", None, "51.0"]

datos_convertidos = [convertir_a_float(valor) for valor in datos_crudos]
datos_convertidos
````

## 10. Buenas prácticas de programación en notebooks

Un notebook de análisis de datos debe ser comprensible y reproducible.

Buenas prácticas:

10. Guardar el trabajo en un repositorio.

````python
def calcular_promedio(valores):

  """
      Calcula el promedio de una lista de valores numéricos.

    Args:
        valores (list): Lista de valores numéricos.

    Returns:
        float: El promedio de los valores.
  """

  if len(valores)==0:
    return None
  return sum(valores)/len(valores)
calcular_promedio([10, 20, 30])
````

## 16. Mini caso de cierre

Aplique funciones, comprensión de listas, manejo de errores y buenas prácticas a un pequeño conjunto de datos crudos.

````python
datos_crudos = ["34.5", "36.0", "", "error", "42.5", None, "51.0"]

def convertir_a_float(valor):
    """
    Utiliza try-except para convierte datos de texto a flotante; datos vaciós y
    datos inválidos son ignorados.
    """
    try:
        return float(valor)
    except (ValueError, TypeError):
        return None

def clasificar_temperatura(temp):
    """
    Clasifica temperaturas según un umbral con condicionales.
    Si es None, devuelve "Dato inválido", si es menor a 40, devuelve "Normal",
    si es menor a 50, devuelve "Precaución", si es mayor o igual a 50, devuelve "Alerta".
    """
    if temp is None:
        return "Dato inválido"
    elif temp < 40:
        return "Normal"
    elif temp < 50:
        return "Precaución"
    else:
        return "Alerta"

# Utiliza compresión de listas para transforma la lista datos_crudos iterando elemento
  # por elemento, aplicando la función convertir_a_float a cada uno para obtener números
  # decimales y guardarlos en una nueva lista.
temperaturas_limpias = [convertir_a_float(valor) for valor in datos_crudos]

# Utiliza compresión de listas para tomar la lista de números ya convertidos
  # (temperaturas_limpias) y le aplica la función clasificar_temperatura a cada uno para
  # generar una nueva lista con sus respectivos estados o categorías.
estados = [clasificar_temperatura(temp) for temp in temperaturas_limpias]


resumen = {
    "datos_originales": datos_crudos,
    "datos_convertidos": temperaturas_limpias,
    "estados": estados
}

resumen
````
