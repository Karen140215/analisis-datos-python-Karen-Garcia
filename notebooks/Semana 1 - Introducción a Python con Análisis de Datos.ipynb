## 4. Variables aplicadas a un contexto de ingeniería

Una variable permite almacenar información. En análisis de datos, las variables pueden representar mediciones, resultados, nombres de equipos, estados de operación, fechas o cualquier dato útil para resolver un problema.

```python
equipo = "Motor Dc"
voltaje_V = 12.0
corriente_A = 1.8

potencia_W = voltaje_V * corriente_A

print("Equipo:", equipo)
print("Voltaje:", voltaje_V, "V")
print("Corriente:", corriente_A, "A")
print("Potencia:", potencia_W, "W")
```

## 5. Listas y cálculo básico

Las listas permiten almacenar varios valores. Por ejemplo, se pueden guardar varias mediciones de temperatura tomadas durante una prueba.

```python
temperaturas_C = [35.2, 36.8, 38.1, 40.5, 43.0, 45.6]

print("Temperaturas:", temperaturas_C)
print("Número de mediciones:", len(temperaturas_C))
print("Temperatura mínima:", min(temperaturas_C))
print("Temperatura máxima:", max(temperaturas_C))
print("Temperatura promedio:", sum(temperaturas_C) / len(temperaturas_C))
```

## 6. Funciones simples

Una función permite reutilizar una operación. En análisis de datos, las funciones ayudan a ordenar el código y evitar repeticiones.

```python
def calcular_potencia(voltaje, corriente):
    potencia = voltaje * corriente
    return potencia

calcular_potencia(12, 1.8)
```

También se pueden usar funciones para clasificar datos según una condición.

```python
def clasificar_temperatura(temperatura):
    if temperatura < 40:
        return "Normal"
    elif temperatura < 45:
        return "Precaucion"
    else:
        return "Alerta"
for temp in temperaturas_C:
    print(temp, "°C ->", clasificar_temperatura(temp))
```

## 7. Primer acercamiento a una tabla de datos con Pandas

En análisis de datos, normalmente trabajamos con tablas. En Python, una de las estructuras más utilizadas para esto es el `DataFrame` de la librería `pandas`.

```python
datos = {
    "muestra": [1, 2, 3, 4, 5, 6],
    "voltaje_V": [12.0, 12.1, 11.9, 12.2, 12.0, 12.3],
    "corriente_A": [1.6, 1.7, 1.8, 2.0, 2.1, 2.3],
    "temperatura_C": [35.2, 36.8, 38.1, 40.5, 43.0, 45.6]
}
df = pd.DataFrame(datos)
df
```

Agreguemos una nueva columna calculada: potencia eléctrica.

```python
df["potencia_W"] = df["voltaje_V"] * df["corriente_A"]
df
```

Exploremos información básica del conjunto de datos.

```python
df.info()
```

```python
df.describe()
```

## 8. Primera visualización

Una gráfica permite observar tendencias y comunicar resultados de manera clara.

```python
plt.figure(figsize=(8, 4))
plt.plot(df["muestra"], df["temperatura_C"], marker="o")
plt.title("Temperatura durante la prueba")
plt.xlabel("Muestra")
plt.ylabel("Temperatura [°C]")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df["muestra"], df["voltaje_V"], marker="o")
plt.title("Voltaje durante la prueba")
plt.xlabel("Muestra")
plt.ylabel("Voltaje [V]")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df["muestra"], df["corriente_A"], marker="o", color="orange")
plt.title("Corriente durante la prueba")
plt.xlabel("Muestra")
plt.ylabel("Corriente [A]")
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 4))
plt.plot(df["muestra"], df["potencia_W"], marker="o", color="green")
plt.title("Potencia durante la prueba")
plt.xlabel("Muestra")
plt.ylabel("Potencia [W]")
plt.grid(True)
plt.show()
```
