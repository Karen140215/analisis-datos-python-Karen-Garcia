  * (1.0) Leer dichos datos y **calcular el promedio** de todos los registros de temperatura, presión y humedad para así obtener un solo valor por variable. Redondear cada registro a solo dos números decimales.
  * (1.0) Sacar los valores máximos y mínimos de cada métrica.

  Tenga en cuenta que debe agregar las unidades de medida correspondientes: grados celsius para temperatura (°C), hectopascales para presión (hPa) y porcentaje para humedad (%).

```python
import random
datos = []
for _ in range(100):
    temperatura = round(random.uniform(20, 30), 2)
    presion = round(random.uniform(1000, 1020), 2)
    humedad = round(random.uniform(50, 70), 2)
    datos.append({
        'temperatura': temperatura,
        'presion': presion,
        'humedad': humedad
    })
```
