1. El código a continuación, genera registros aleatorios de métricas de temperatura, presión y humedad los cuales se guardan en la variable `datos`. Su trabajo consiste en **desarrollar funciones** en python que haga lo siguiente:

  * (1.0) Leer dichos datos y **calcular el promedio** de todos los registros de temperatura, presión y humedad para así obtener un solo valor por variable. Redondear cada registro a solo dos números decimales.
  * (1.0) Sacar los valores máximos y mínimos de cada métrica.

  Tenga en cuenta que debe agregar las unidades de medida correspondientes: grados celsius para temperatura (°C), hectopascales para presión (hPa) y porcentaje para humedad (%).\

```python
import random
datos = []
for _ in range(100):
    temperatura = round(random.uniform(20, 30), 2)
    presion = round(random.uniform(1000, 1020), 2)
    humedad = round(random.uniform(50, 70), 2)
    datos.append({
        'temperatura': temperatura,
        'presion': presion,
        'humedad': humedad
    })
```

```python
def calcular_promedio(datos):

    n = len(datos)
    promedio = {
        'Temperatura (°C)': round(sum(reg['temperatura'] for reg in datos)/n),
        'Presión (hPa)': round(sum(reg['presion'] for reg in datos)/n),
        'Humedad (%)': round(sum(reg['humedad'] for reg in datos)/n),
    }
    return promedio


def calcular_maximos_minimos(datos):

    maximos = {
        'Temperatura (°C)': max(reg['temperatura'] for reg in datos),
        'Presión (hPa)': max(reg['presion'] for reg in datos),
        'Humedad (%)': max(reg['humedad'] for reg in datos),
    }

    minimos = {
        'Temperatura (°C)': min(reg['temperatura'] for reg in datos),
        'Presión (hPa)': min(reg['presion'] for reg in datos),
        'Humedad (%)': min(reg['humedad'] for reg in datos),
    }
    return maximos, minimos

promedio = calcular_promedio(datos)
maximos, minimos = calcular_maximos_minimos(datos)

```
