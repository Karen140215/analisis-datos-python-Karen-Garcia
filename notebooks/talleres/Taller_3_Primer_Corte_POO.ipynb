## Punto 1 – Diseño orientado a objetos y abstracción (1.0)

Diseñe una clase abstracta `Sensor` que represente un sensor genérico.
- Atributo protegido `_id_sensor`
- Atributo protegido `_unidad`
- Método abstracto `leer()` que retorne una medición
- Método concreto `info()` que retorne una cadena con el ID y la unidad

Implemente dos clases concretas `SensorTemperatura` y `SensorPresion`, cada una sobrescribiendo `leer()` con un valor simulado.

```python
from abc import ABC, abstractmethod
import random

# Clase abstracta
class Sensor(ABC):
    def __init__(self, id_sensor, unidad):
        self._id_sensor = id_sensor
        self._unidad = unidad

    @abstractmethod
    def leer(self):
        pass

    def info(self):
        return f"ID: {self._id_sensor}, Unidad: {self._unidad}"

# Clases concretas
class SensorTemperatura(Sensor):
    def leer(self):
        return round(random.uniform(15.0, 30.0), 1)

class SensorPresion(Sensor):
    def leer(self):
        return round(random.uniform(900.0, 1000.0), 1)

# Prueba simple
temp = SensorTemperatura("T01", "°C")
pres = SensorPresion("P01", "hPa")

print(temp.info(), "-> Lectura:", temp.leer())
print(pres.info(), "-> Lectura:", pres.leer())
```

## Punto 2 – Encapsulamiento y validación con propiedades (1.0)

Cree una clase `CuentaBancaria` con:
- Atributo privado `__saldo`
- Métodos `depositar(monto)` y `retirar(monto)`
- Propiedad `saldo` de solo lectura

Condiciones:
- No se permite retirar más dinero del saldo disponible
- No se permiten montos negativos

```python
class CuentaBancaria:
    def __init__(self, saldo_inicial=0):
        # Atributo privado con doble guion bajo
        # Si el saldo inicial es negativo, max() lo convierte a 0 para no permitir montos negativos, es un comparador de numeros, toda el mayor.
        self.__saldo = max(0, saldo_inicial)

    # property convierte este método en una propiedad de SOLO LECTURA.
    # Permite ver el valor de __saldo sin paréntesis, pero bloquea asignaciones como 'cuenta.saldo = 500'
    @property
    def saldo(self):
        return self.__saldo

    def depositar(self, monto):
        if monto > 0:
            self.__saldo += monto
            print(f"Depositados: ${monto}")
        else:
            print("El monto a depositar debe ser mayor a 0")

    def retirar(self, monto):
        if monto <= 0:
            print("El monto a retirar debe ser mayor a 0")
        elif monto > self.__saldo:
            print("Saldo insuficiente")
        else:
            self.__saldo -= monto
            print(f"Retirados: ${monto}")

cuenta = CuentaBancaria(100)

cuenta.depositar(50)   # Suma 50 -> Saldo: 150
cuenta.retirar(30)     # Resta 30 -> Saldo: 120
cuenta.retirar(200)    # Error: Saldo insuficiente
cuenta.depositar(-10)  # Error: Monto inválido

print("Saldo final:", cuenta.saldo)
```

## Punto 3 – Polimorfismo y listas de objetos (1.0)

Dada una lista de objetos de tipo `Sensor` (temperatura y presión), implemente una función `leer_sensores(lista_sensores)` que:
- Recorra la lista
- Llame al método `leer()` de cada objeto
- Imprima la información del sensor y su lectura

```python
def leer_sensores(lista_sensores):
    # Recorre la lista de sensores recibida por parámetro
    for sensor in lista_sensores:
        # Obtiene la lectura invocando el método leer() de cada sensor (Polimorfismo)
        lectura = sensor.leer()

        # Imprime la información básica y el valor medido con su unidad
        print(f"{sensor.info()} Lectura actual: {lectura} {sensor._unidad}")

# Creación de la lista de objetos Sensor
mis_sensores = [
    SensorTemperatura("TEMP-01", "°C"),
    SensorPresion("PRES-01", "hPa"),
    SensorTemperatura("TEMP-02", "°C"),
    SensorPresion("PRES-02", "PSI")
]

# Ejecución de la función
leer_sensores(mis_sensores)
```

## Punto 4 – Composición vs herencia (1.0)
### Enunciado
Implemente una clase `SistemaMonitoreo` que:
- Contenga una lista de sensores (composición)
- Tenga un método `agregar_sensor(sensor)`
- Tenga un método `reporte()` que muestre todas las lecturas

Justifique brevemente por qué aquí es mejor composición que herencia.

```python
class SistemaMonitoreo:
    def __init__(self):
        # La lista vive DENTRO del sistema (Composición)
        self.sensores = []

    def agregar_sensor(self, sensor):
        # Añade un nuevo objeto sensor a la lista
        self.sensores.append(sensor)

    def reporte(self):
        # Muestra las lecturas de todos los sensores registrados
        print("=== REPORTE DEL SISTEMA DE MONITOREO ===")
        for sensor in self.sensores:
            print(f"{sensor.info()} Lectura actual: {sensor.leer()} {sensor._unidad}")

# 1. Creamos el sistema
sistema = SistemaMonitoreo()

# 2. Agregamos sensores al sistema
sistema.agregar_sensor(SensorTemperatura("TEMP-01", "°C"))
sistema.agregar_sensor(SensorPresion("PRES-01", "hPa"))
sistema.agregar_sensor(SensorPresion("PRES-02", "hPa"))
sistema.agregar_sensor(SensorTemperatura("TEMP-02", "°C"))

sistema.reporte()
```

Es mejor utilizar composición porque responde a la relación lógica del mundo real: un sistema monitoreo tiene sensores, no es un sensor. Si usáramos herencia estaríamos diciendo erróneamente que el sistema en sí mismo es un dispositivo de medición. En cambio la composición nos permite tratar al sistema como un contenedor flexible que puede almacenar, organizar y consultar cualquier cantidad y tipo de sensores dentro de una lista sin alterar la estructura original de las clases.


## Punto 5 – Gestión de estados y responsabilidad única (SRP) (1.0)

Diseñe una clase `Experimento` que represente un experimento computacional o de laboratorio.

Requisitos:
1. Atributos: `nombre` y `estado` (solo: `'creado'`, `'configurado'`, `'ejecutado'`).
2. Métodos:
   - `configurar()`: solo si el estado es `'creado'`
   - `ejecutar()`: solo si el estado es `'configurado'`
   - `reiniciar()`: vuelve a `'creado'`
3. Si una acción no está permitida, lanzar una excepción con mensaje claro.
4. Método `info()` que muestre nombre y estado.

```python
class Experimento:
    def __init__(self, nombre):
        self.nombre = nombre
        self.estado = 'creado'

    def configurar(self):
        if self.estado != 'creado':
            raise Exception("Error: Solo se puede configurar si el estado es 'creado'")
        self.estado = 'configurado'
        print("Experimento configurado.")

    def ejecutar(self):
        if self.estado != 'configurado':
            raise Exception("Error: Solo se puede ejecutar si el estado es 'configurado'")
        self.estado = 'ejecutado'
        print("Experimento ejecutado.")

    def reiniciar(self):
        self.estado = 'creado'
        print("Experimento reiniciado a 'creado'.")

    def info(self):
        return f"Experimento: {self.nombre} | Estado: {self.estado}"


# --- Prueba en Colab ---
exp = Experimento("Prueba de Laboratorio")
print(exp.info()) # Estado: creado

exp.configurar()  # Pasa a 'configurado'
exp.ejecutar()    # Pasa a 'ejecutado'
print(exp.info())

exp.reiniciar()   # Vuelve a 'creado'
print(exp.info())

# Si intentas ejecutarlo directamente sin configurar, lanzará la excepción:
#exp.ejecutar()
```
